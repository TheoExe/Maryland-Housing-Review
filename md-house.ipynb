Python notebook to analyze MD Housing Rehab Program loans. Elements analyzed include:
- How many loans were given to a county, for each year (Rate)
- The total amount ($) loaned to individuals for each county, by year (Annual Total)

```python
import pandas
import matplotlib.pyplot as plt
import numpy as np

house = pandas.read_csv("[CLEANED] Maryland_Housing_Rehab_Special_Loans_Program_FY_2011-2023.csv")
popden = pandas.read_csv("[CLEANED] Maryland_Resident_Population_Per_Square_Mile__2010-2024.csv")
```

```python
# Investigating the average number of loans issued in Baltimore City (by fiscal year)
balt_house = house[house['County'] == 'Baltimore City'].sort_values(by='Fiscal Year', ascending=True)

print(balt_house)

print(f"\nAverage number of loans issued in Baltimore City: {round(balt_house['Loans Issued'].mean())}")

# Plotting the loans issued in Baltimore City by fiscal year
plt.plot(balt_house['Fiscal Year'], balt_house['Loans Issued'], marker='o')
plt.title("Baltimore City Rehab Loans Issued by Fiscal Year")
plt.xlabel("Fiscal Year")
plt.ylabel("Loans Issued")
plt.grid()
plt.show()
```

```python
# Investigating the average number of loans issued in Baltimore County (by fiscal year)
baltco_house = house[house['County'] == 'Baltimore'].sort_values(by='Fiscal Year', ascending=True)

print(f"\nAverage number of loans issued in Baltimore County: {round(baltco_house['Loans Issued'].mean())}")

# Plotting the loans issued in Baltimore County by fiscal year
plt.plot(baltco_house['Fiscal Year'], baltco_house['Loans Issued'], marker='o')
plt.title("Baltimore County Rehab Loans Issued by Fiscal Year")
plt.xlabel("Fiscal Year")
plt.ylabel("Loans Issued")
plt.grid()
plt.show()
```

```python
total_loan = house['Loans Issued'].sum() #total loans issued in Maryland

balt_loan = house[house['County'] == 'Baltimore City']['Loans Issued'].sum() #total loans issued in Baltimore City


print(f"Total Baltimore City Rehab Loans Issued (2011 - 2023): {balt_loan}")
print(f"\nTotal Maryland Rehab Loans Issued (2011 - 2023): {total_loan}")
print(f"\nPercent of Baltimore City rehab loans issued vs Maryland total (2011 - 2023): {round((balt_loan/total_loan)*100,2)}%")
```

```python
# Investigating the average number of loans issued in Prince George's County (by fiscal year)
pg_house = house[house['County'] == 'Prince George\'s'].sort_values(by='Fiscal Year', ascending=True)

print(f"\nAverage number of loans issued in Prince George's County: {round(pg_house['Loans Issued'].mean())}")

# Plotting the loans issued in Prince George's County by fiscal year
plt.plot(pg_house['Fiscal Year'], pg_house['Loans Issued'], marker='o')
plt.title("Prince George's County Rehab Loans Issued by Fiscal Year")
plt.xlabel("Fiscal Year")
plt.ylabel("Loans Issued")
plt.grid()
plt.show()
```

```python
color_wheel = ['b', 'g', 'r', 'c', 'm', 'y', 'k']
```

```python
# General means of sorting through County vs Loans Issued
county_names = house["County"].unique() #key value list of county names

# Group counties by name and calculate average loans issued, then sort descending
avg = house.groupby('County')['Loans Issued'].mean().sort_values(ascending=False)   
avg
```

Loans Awarded by County (FY 2011-2023): High Award Rate

```python
x = 0 #iterate through county names
n = 0 #iterate through subplot rows

color_w = 0
tracker = 0 #iterate through award rates


fig, axs = plt.subplots(2,3, figsize=(19,8))
fig.subplots_adjust(wspace=0.3, hspace=0.5)
fig.suptitle('Maryland Counties - Number of Individual Loans Awarded by Fiscal Year (2011-2023): High Award Rate', fontsize=16)

# group county records by max loans issued
max = house.groupby('County')['Loans Issued'].max().sort_values(ascending=False) 

for county in max.index:
    county_data = house[house['County'] == county]
    if county != 'Multiple':
        if x < 3 and tracker < 6:
            #Create bar graph for each county
            axs[n,x].bar(round(county_data['Fiscal Year']), county_data['Loans Issued'], color = color_wheel[tracker])
            axs[n,x].set_title(f"{county}", weight='bold')
            axs[n,x].set_ylabel("Loans Issued")
            axs[n,x].set_xlabel("Fiscal Year")
            axs[n,x].set_xticks(np.arange(2011, 2024, 2))
            x = x + 1

            if x == 3:
                n = n + 1
                x = 0
            tracker = tracker + 1
            if color_w < 6:
                color_w = color_w + 1
            else:
                color_w = 0
    
     
plt.plot()
```

Loan Program Award Totals - Maryland Counties (FY 2011-2023): High Annual Total

```python
x = 0 #iterate through county names
n = 0 #iterate through subplot rows
tracker = 0 #iterate through award rates

color_w = 4

fig, axs = plt.subplots(2,3,figsize=(16,8))
fig.suptitle('Loan Program Award Totals - Maryland Counties (FY 2011-2023): High Annual Total', fontsize=16)
fig.subplots_adjust(wspace=0.5, hspace=0.5)

for county in max_award.index:
    county_data = house[house['County'] == county]
    if county != 'Multiple':
        if x < 3 and tracker < 6:
            #Create bar graph for each county
            axs[n,x].bar(round(county_data['Fiscal Year']), county_data['Award Amount'], color = color_wheel[color_w])
            axs[n,x].set_title(f"{county}", weight='bold')
            axs[n,x].set_ylabel("Total Awarded ($)", va = 'bottom')
            axs[n,x].yaxis.set_label_coords(-0.17,0.5)
            axs[n,x].xaxis.set_label_coords(0.5,-0.17)
            axs[n,x].set_xlabel("Fiscal Year")
            axs[n,x].set_xticks(np.arange(2011, 2025, 3))
            axs[n,x].yaxis.set_major_formatter((lambda x, pos: '{:,.1f}'.format(x/1000000) + 'M'))
            x = x + 1

            if x == 3:
                n = n + 1
                x = 0
            
            if color_w < 5:
                color_w = color_w + 1
            else:
                color_w = 0
        tracker = tracker + 1
    

plt.plot()
```
